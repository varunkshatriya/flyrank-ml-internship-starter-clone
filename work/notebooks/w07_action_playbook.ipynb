## 1. Ranked actions + reason codes

## 1. Ranked actions + reason codes

The playbook turns the observed search signals from the baseline/model workflow into a prioritized review queue.

| Priority | Action | Reason code | When to use |
|---|---|---|---|
| High | Review CTR and search snippet | LOW_CTR | CTR is weak relative to comparable position performance |
| High | Review content freshness | STALE_CONTENT | Content age suggests the page may need a freshness review |
| Medium | Review search demand | HIGH_VOLUME_OPPORTUNITY | The query has meaningful search volume and may justify improvement work |
| Medium | Review page position and CTR together | CTR_POSITION_MISMATCH | CTR appears weaker than expected for the observed position |
| Low | Monitor without immediate change | WEAK_SIGNAL | Signals are not strong enough to justify an immediate content action |

The ranking is intended for human review. A high score means higher priority for investigation, not an automatic instruction to change content.

## 2. Intended use and limits

*Who uses this, for what — and where it stops being valid.*

## 2. Intended use and limits

### Intended use

This playbook is intended to help a content or SEO analyst prioritize pages for manual review.

It can help answer:

- Which pages should be reviewed first?
- Which observed search signals explain the recommendation?
- What type of investigation should happen next?

### Limits

The recommendations are decision-support only.

The playbook does not prove that a page is causing a ranking decline, and it does not predict Google's ranking decisions.

The signals are based on the available dataset and observed measurements. Search intent, SERP changes, competitors, technical issues, content quality, seasonality, and measurement noise may not be fully represented.

The recommendations should therefore be reviewed by a person before any content or SEO change is made.

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

## 3. Human review + the no-go list

### Human review required before acting

A person should check:

1. Whether the page matches the current search intent.
2. Whether the observed CTR and position are plausible for the query.
3. Whether the content is actually outdated or incomplete.
4. Whether there are important SERP changes that the dataset does not capture.
5. Whether the recommended action could negatively affect useful existing content.
6. Whether the recommendation is supported by more than one weak signal.

### No-go list

The system should never automatically:

- Delete a page.
- Rewrite or publish content without human approval.
- Change factual claims without verification.
- Change canonical tags or indexing settings automatically.
- Redirect a page automatically.
- Make decisions based only on a model score.
- Treat a prediction as proof of a ranking cause.

The model and playbook are therefore prioritization tools, not autonomous content decision-makers.

## 4. Monitoring / retrain triggers

*What would tell you the recommendations went stale?*

## 4. Monitoring / retrain triggers

The recommendations should be reviewed when the underlying search behavior changes.

### Monitoring triggers

Reconsider the playbook when:

- CTR distributions change substantially.
- Search-volume distributions change substantially.
- Average-position distributions shift.
- The relationship between CTR and position changes.
- The proportion of declining pages changes substantially.
- The model's Precision@50 falls below the Week-4 baseline.
- New data fields or tracking definitions are introduced.

### Retraining trigger

Retraining should be considered when new labelled observations are available and the current model consistently performs below the simple baseline on an equivalent evaluation design.

A retrain should use a fresh validation split and should repeat the leakage audit.

A change in model performance should be treated as an observed signal to investigate rather than an automatic reason to deploy a new model.
