## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

## Research Question

**Decision**

Which webpages should be prioritized for SEO improvements?

**Who acts?**

SEO specialists or content editors.

**Action**

Improve the highest-priority pages by updating content, titles, metadata, internal links, or technical SEO.

**Unit of analysis**

One webpage (one content item).

**Output**

A ranking score that prioritizes webpages for optimization.

**Task type**

Ranking / Scoring.

**Success metric**

Precision@K (evaluated later in the project).

**Cost of a wrong recommendation**

If a low-priority page is recommended first, the team wastes time optimizing pages with little impact.

If an important page is missed, valuable organic traffic and business opportunities may be lost.

**Why ML instead of a rule?**

Search performance depends on many interacting signals such as CTR, impressions, clicks, engagement, position, and content characteristics. These relationships are too complex for a single fixed rule, making machine learning a useful decision-support tool.
